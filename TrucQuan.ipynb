```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os

# =========================================================
# CẤU HÌNH
# =========================================================
# Đường dẫn đến file CSV confusion matrix bạn đã lưu trước đó
# Lưu ý: Cập nhật đường dẫn nếu file nằm ở thư mục khác
csv_path = "./phobert-ft-tuned/confusion_matrix.csv"
output_image_path = "./phobert-ft-tuned/confusion_matrix.png"

# =========================================================
# 1. ĐỌC DỮ LIỆU
# =========================================================
if os.path.exists(csv_path):
    # index_col=0: Để pandas biết cột đầu tiên là tên nhãn (Negative, Neutral, Positive)
    df_cm = pd.read_csv(csv_path, index_col=0)
    print("Dữ liệu Confusion Matrix:")
    print(df_cm)
else:
    print(f"Lỗi: Không tìm thấy file tại {csv_path}. Vui lòng kiểm tra lại đường dẫn.")
    exit()

# =========================================================
# 2. VẼ BIỂU ĐỒ (HEATMAP)
# =========================================================
plt.figure(figsize=(10, 8))

# Sử dụng Seaborn heatmap
# annot=True: Hiển thị con số cụ thể trong từng ô
# fmt='d': Định dạng số nguyên (decimal), tránh hiển thị dạng khoa học (1e2)
# cmap='Blues': Tông màu xanh (đậm dần theo giá trị)
heatmap = sns.heatmap(
    df_cm, 
    annot=True, 
    fmt='d', 
    cmap='Blues', 
    linewidths=.5, 
    annot_kws={"size": 14} # Cỡ chữ số
)

# =========================================================
# 3. TRANG TRÍ VÀ LƯU ẢNH
# =========================================================
plt.title('Confusion Matrix - PhoBERT Model', fontsize=18, pad=20)
plt.ylabel('Thực tế (Actual Label)', fontsize=14)
plt.xlabel('Dự đoán (Predicted Label)', fontsize=14)

# Chỉnh sửa tick labels (nhãn trục) cho dễ đọc hơn nếu cần
plt.xticks(fontsize=12, rotation=45)
plt.yticks(fontsize=12, rotation=0)

# Lưu hình ảnh
plt.tight_layout() # Tự động căn chỉnh lề
plt.savefig(output_image_path, dpi=300) # dpi=300 để ảnh sắc nét in ấn được
print(f"\nĐã lưu biểu đồ thành công tại: {output_image_path}")

# Hiển thị (nếu chạy trên Jupyter Notebook/Local)
plt.show()
```
